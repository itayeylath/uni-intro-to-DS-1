```python
import sqlite3
import pandas as pd
```

```python
# Question 1
q1_1 = """
SELECT Name, Country
FROM employees
WHERE Salary > 5000;
"""

q1_2 = """
SELECT Name
FROM employees
ORDER BY Experience DESC
LIMIT 1;
"""

q1_3 = """
SELECT Country, COUNT(*) AS EmployeeCount
FROM employees
GROUP BY Country;
"""

q1_4 = """
SELECT AVG(Salary) AS AvgSalary
FROM employees
WHERE Experience >= 12;
"""

q1_5 = """
SELECT Country
FROM employees
ORDER BY Salary DESC
LIMIT 1;
"""
```

#

Create the `employees` table, insert data, and execute the Question 1 queries.


```python
# Question 2
conn = sqlite3.connect(":memory:")
cur = conn.cursor()

cur.execute(
    """
    CREATE TABLE employees (
        Name TEXT,
        Experience INTEGER,
        Salary INTEGER,
        Country TEXT
    );
    """
)

employees_data = [
    ("Alice", 23, 8000, "Germany"),
    ("Barbara", 17, 7200, "France"),
    ("Camila", 11, 7500, "Spain"),
    ("Danna", 54, 8800, "Portugal"),
    ("Evan", 32, 9100, "Italy"),
    ("Fatima", 28, 7900, "Germany"),
    ("Hugo", 14, 6400, "Spain"),
]

cur.executemany(
    "INSERT INTO employees (Name, Experience, Salary, Country) VALUES (?, ?, ?, ?);",
    employees_data,
)
conn.commit()


def run_query(sql: str) -> pd.DataFrame:
    return pd.read_sql_query(sql, conn)
```

```python
print(q1_1.strip())
q1_1_df = run_query(q1_1)
q1_1_df
```

```python
# Question 3
q3_1 = """
SELECT e.Name
FROM employees e
JOIN countries c ON e.Country = c.Name
WHERE c.Population >= 50;
"""

q3_2 = """
SELECT c.Name AS Country, AVG(e.Experience) AS AvgExperience
FROM employees e
JOIN countries c ON e.Country = c.Name
WHERE c.Population <= 80
GROUP BY c.Name;
"""

q3_3 = """
SELECT c.Name AS Country, AVG(e.Salary) AS AvgSalary, c.Population
FROM employees e
JOIN countries c ON e.Country = c.Name
GROUP BY c.Name, c.Population;
"""

q3_4 = """
SELECT e.Name, e.Experience
FROM employees e
JOIN countries c ON e.Country = c.Name
ORDER BY c.Population DESC;
"""

q3_5 = """
SELECT e.Name
FROM employees e
JOIN countries c ON e.Country = c.Name
WHERE c.Population >= 20
  AND c.Name <> 'Germany'
  AND e.Experience >= 24;
"""
```
