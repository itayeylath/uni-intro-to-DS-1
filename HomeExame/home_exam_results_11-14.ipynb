```python
# Question 11 A

import csv

file_path = "csv_file"


def get_USD_rate():
    return ""  # given in the exam


def read_transactions(path):
    rows = []
    with open(path, "r", encoding="utf-8") as f:
        reader = csv.reader(f)
        for row in reader:
            if not row:
                continue
            row = [cell.strip() for cell in row]
            if len(row) < 4:
                continue
            date_str, account_id, amount_str, currency = row[:4]
            try:
                amount = float(amount_str)
            except ValueError:
                continue
            rows.append((date_str, account_id, amount, currency))
    return rows


transactions = read_transactions(file_path)

net_balance_usd = 0.0
for date_str, account_id, amount, currency in transactions:
    if currency == "USD":
        usd_amount = amount
    else:
        usd_amount = amount / get_USD_rate(currency)
    net_balance_usd += usd_amount

print(f"Final balance (USD): {net_balance_usd}")

# Question 11B

currency_min = {}
currency_max = {}

for _, _, amount, currency in transactions:
    if currency not in currency_min:
        currency_min[currency] = amount
        currency_max[currency] = amount
    else:
        currency_min[currency] = min(currency_min[currency], amount)
        currency_max[currency] = max(currency_max[currency], amount)

for currency in sorted(currency_min):
    print(
        f"Currency: {currency} | Highest income: {currency_max[currency]} | "
        f"Highest expense: {currency_min[currency]}"
    )

# Question 11C

daily_income = {}
daily_expense = {}

for date_str, _, amount, _ in transactions:
    if date_str not in daily_income:
        daily_income[date_str] = 0.0
        daily_expense[date_str] = 0.0

    if amount > 0:
        daily_income[date_str] += amount
    elif amount < 0:
        daily_expense[date_str] += amount

num_days = len(daily_income)

avg_daily_income = sum(daily_income.values()) / num_days if num_days else 0.0
avg_daily_expense = sum(daily_expense.values()) / num_days if num_days else 0.0

print(f"Average daily income: {avg_daily_income}")
print(f"Average daily expense: {avg_daily_expense}")
```

```python
# Question 13A

import pandas as pd

df_grades = pd.read_csv("grased.csv", header=None, names=["student_id", "course_id", "grade"])
df_grades["grade"] = pd.to_numeric(df_grades["grade"], errors="coerce")

grouped = df_grades.groupby("course_id")["grade"].mean().dropna()

for course_id, avg in grouped.items():
    print(f"{course_id}, {avg:.2f}")

# Question 13B

lowest_course = grouped.idxmin()
highest_course = grouped.idxmax()

print(f"Lowest average: {lowest_course}, {grouped.loc[lowest_course]:.2f}")
print(f"Highest average: {highest_course}, {grouped.loc[highest_course]:.2f}")

# Question 13C

course_counts = df_grades.groupby("student_id")["course_id"].nunique()

student_id = course_counts.idxmax()
num_courses = int(course_counts.loc[student_id])

print(f"Student with most courses: {student_id}, {num_courses}")

```
