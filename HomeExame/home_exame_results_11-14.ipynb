```python
# Question 11 A

import csv
file_path = "csv_file"

def get_USD_rate():
    return "" # Assuming it's return the rate of the currency to USD relate to the ILS or EUR

net_balance_usd = 0.0

with open(file_path, "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        if not row:
            continue
        row = [cell.strip() for cell in row]
        if len(row) < 4:
            continue

        date_str, account_id, amount_str, currency = row[0], row[1], row[2], row[3]
        try:
            amount = float(amount_str)
        except ValueError:
            continue

        if currency == "USD":
            usd_amount = amount
        else:
            usd_amount = amount / get_USD_rate(currency)

        net_balance_usd += usd_amount


# Question 11B

currency_min = {}
currency_max = {}

with open(file_path, "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        if not row:
            continue
        row = [cell.strip() for cell in row]
        if len(row) < 4:
            continue

        amount_str, currency = row[2], row[3]
        try:
            amount = float(amount_str)
        except ValueError:
            continue

        if currency not in currency_min:
            currency_min[currency] = amount
            currency_max[currency] = amount
        else:
            if amount < currency_min[currency]:
                currency_min[currency] = amount
            if amount > currency_max[currency]:
                currency_max[currency] = amount

for currency in sorted(currency_min):
    print(
        f"Currency: {currency} | Highest income: {currency_max[currency]} | "
        f"Highest expense: {currency_min[currency]}"
    )

# Question 11C

daily_income = {}
daily_expense = {}

with open(file_path, "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        if not row:
            continue
        row = [cell.strip() for cell in row]
        if len(row) < 4:
            continue

        date_str, amount_str = row[0], row[2]
        try:
            amount = float(amount_str)
        except ValueError:
            continue

        if date_str not in daily_income:
            daily_income[date_str] = 0.0
            daily_expense[date_str] = 0.0

        if amount > 0:
            daily_income[date_str] += amount
        elif amount < 0:
            daily_expense[date_str] += amount

num_days = len(daily_income)

avg_daily_income = sum(daily_income.values()) / num_days if num_days else 0.0
avg_daily_expense = sum(daily_expense.values()) / num_days if num_days else 0.0

print(f"Average daily income: {avg_daily_income}")
print(f"Average daily expense: {avg_daily_expense}")
```
